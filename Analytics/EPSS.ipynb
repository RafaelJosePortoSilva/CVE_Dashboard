```python
import requests
import pandas as pd
```

```python
def get_epss_historical_data():
    """
    Fetches the EPSS (Exploit Prediction Scoring System) Historical data from the First.org API.

    Returns:
        list: A list of dictionaries containing the EPSS data.
    """
    response = requests.get('https://api.first.org/data/v1/epss')
    if response.status_code == 200:
        return response.json()['data']
    else:
        raise Exception(f"Failed to fetch data: {response.status_code}")

def process_epss_data(data):
    """
    Processes the EPSS data into a pandas DataFrame.

    Args:
        data (list): A list of dictionaries containing the EPSS data.

    Returns:
        pd.DataFrame: A DataFrame containing the processed EPSS data.
    """
    df = pd.DataFrame(data)
    df['date'] = pd.to_datetime(df['date'])
    return df
```

```python
process_epss_data(get_epss_historical_data())
```
